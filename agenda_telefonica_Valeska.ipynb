```python
class Contacto:
    """Representa un contacto con nombre, teléfono(s) y correo."""
    def __init__(self, nombre, telefono, mail):
        # Propiedades de instancia
        self.nombre = nombre
        self.telefono = telefono
        self.mail = mail

    def __str__(self):
        # Cadena de presentación detallada
        return (f"\n--- Contacto: {self.nombre} ---\n"
                f"Teléfono : {self.telefono}\n"
                f"Mail: {self.mail}")

class Agenda:
    """Maneja la colección de objetos Contacto."""
    def __init__(self):
        # El diccionario guardará los contactos, usando el nombre como clave
        self.contactos = {}

    def agregar_contacto(self, nombre, telefono="", mail=""):
        """Agrega un nuevo contacto a la agenda."""
        # Se normaliza el nombre a minúsculas para evitar duplicados por mayúsculas
        clave = nombre.lower()
        if clave in self.contactos:
            print(f"❌ Error: El contacto '{nombre}' ya existe.")
            return

        # Crea la instancia de Contacto y la guarda en el diccionario
        nuevo_contacto = Contacto(nombre, telefono, mail)
        self.contactos[clave] = nuevo_contacto
        print(f"✅ Contacto '{nombre}' agregado con éxito.")

    def buscar_contacto(self, nombre):
        """Busca un contacto por nombre y lo devuelve."""
        clave = nombre.lower()
        return self.contactos.get(clave)

    def desplegar_contacto(self, nombre):
        """Despliega la información completa de un contacto."""
        contacto = self.buscar_contacto(nombre)
        if contacto:
            # Llama automáticamente al método __str__ de la clase Contacto
            print(contacto)
        else:
            print(f"🔍 Contacto '{nombre}' no encontrado.")

    def eliminar_contacto(self, nombre):
        """Elimina un contacto de la agenda."""
        clave = nombre.lower()
        if clave in self.contactos:
            del self.contactos[clave]
            print(f"🗑️ Contacto '{nombre}' eliminado con éxito.")
        else:
            print(f"🔍 Contacto '{nombre}' no encontrado, no se puede eliminar.")

def main():
    """Función principal para demostrar el uso de la Agenda."""
    mi_agenda = Agenda()

    print("--- 1. AGREGANDO CONTACTOS ---")
    # Agregar Contacto (incluyendo telefono2)
    mi_agenda.agregar_contacto("Matias", "997157488", "matiiasjaavier@gmai.com")
    mi_agenda.agregar_contacto("Valeska", "946979451", "vale92hennerstoledo@gmail.com")
    mi_agenda.agregar_contacto("Pedro","94795522","pedro.g@gmail.com")


    print("\n--- 2. DESPLEGANDO UN CONTACTO ---")
    # Desplegar la información de un contacto
    mi_agenda.desplegar_contacto("Matias")

    print("\n--- 3. BUSCANDO UN CONTACTO (No se despliega automáticamente, solo se verifica) ---")
    # Buscar un contacto por nombre
    busqueda = mi_agenda.buscar_contacto("Valeska")
    if busqueda:
        print(f"✔️ Contacto encontrado: {busqueda.nombre}")

    print("\n--- 4. ELIMINANDO UN CONTACTO ---")
    # Eliminar un contacto
    mi_agenda.eliminar_contacto("Pedro")

    print("\n--- 5. INTENTO DE DESPLEGAR EL CONTACTO ELIMINADO ---")
    mi_agenda.desplegar_contacto("Pedro")

# Ejecutar la función principal
if __name__ == "__main__":
    main()
```
